```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set a Seaborn style correctly
sns.set_theme(style="darkgrid")  # Use other styles like "whitegrid", "dark", "ticks"

# Sample plot to test
import numpy as np
import pandas as pd

# Generate sample data
data = pd.DataFrame({
    "x": np.linspace(0, 10, 100),
    "y": np.sin(np.linspace(0, 10, 100))
})

# Create a plot
plt.figure(figsize=(8, 5))
sns.lineplot(x="x", y="y", data=data)
plt.show()
```
